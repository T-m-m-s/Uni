```python
def filtra_personaggi1(d, e, h):
    temp4 = d[(d['eye_color'] == e) & (d['hair_color'] == h)]
    eighteen = temp4
    display(eighteen)

filtra_personaggi1(df, 'blue', 'blond')
```

```python
def filtra_personaggi2(d, e, h):
    valid_hair = df['hair_color'].unique()
    valid_eye = df['eye_color'].unique()
    if h not in valid_hair:
        print(f"Il colore dei capelli '{h}' non presente nel DataFrame")
        return
    if e not in valid_eye:
        print(f"Il colore degli occhi '{e}' non presente nel DataFrame")
        return
    nineteen = d[(d['eye_color'] == e) & (d['hair_color'] == h)]
    if not nineteen.empty:
        display(nineteen)
    else:
        print("Non ci sono personaggi che soddisfano entrambi i criteri")

filtra_personaggi2(df, 'blue', 'blond')
```

```python
df_movies = pd.DataFrame({})
for index, films in df[['films']].itertuples():
    for film in films.split(','):
        film = film.strip()
        if film:
            df_movies = df_movies._append({'id_character': index,'title': film}, ignore_index=True)

df_movies.set_index('id_character', inplace=True)

filter_1 = df_movies[df_movies.index == 0]

filter_2 = df_movies[df_movies['title'] == 'Return of the Jedi']

twenty_1 = pd.merge(df['name'], filter_1, left_index=True, right_index=True)
display(twenty_1)

twenty_2 = pd.merge(df['name'], filter_2, left_index=True, right_index=True)
display(twenty_2)
```

```python
df_vehicles = pd.DataFrame({})
for index, vehicles in df[['vehicles']].itertuples():
    if isinstance(vehicles, str):
        for vehicle in vehicles.split(','):
            vehicle = vehicle.strip()
            if vehicle:
                df_vehicles = df_vehicles._append({'id_character': index, 'vehicle': vehicle}, ignore_index=True)
    else:
        pass

df_vehicles.set_index('id_character', inplace=True)

twentyone_2 = pd.merge(df[df['name'] == 'Anakin Skywalker']['name'], df_vehicles, left_index=True, right_index=True)
display(twentyone_2)

filter_3 = df_vehicles[df_vehicles['vehicle'] == 'Tribubble bongo']

twentyone_3 = pd.merge(df['name'], filter_3, left_index=True, right_index=True)
display(twentyone_3)

filter_5 = df
filter_5.loc[filter_5['vehicles_count'].idxmax()]
filter_5 = filter_5['name']

twentyone_5 = pd.merge(filter_5, df_vehicles, left_index=True, right_index=True)
```
