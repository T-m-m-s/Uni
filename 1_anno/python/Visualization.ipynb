```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('seaborn-v0_8-whitegrid')
x = np.linspace(0, 10, 100)
fig = plt.figure( figsize=(8, 2) )
plt.plot(x, np.sin(x), '-')
plt.plot(x, np.cos(x), '--');
```

```python
plt.figure( figsize = (8, 4) ) # Creiamo una nuova figura 

plt.subplot(2, 1, 1) # (righe, colonne, numero di pannello ) 
plt.plot(x, np.sin(x))

plt.subplot(2, 1, 2) # (righe, colonne, numero di pannello ) 
plt.plot(x, np.cos(x));
```

```python
# Creiamo la griglia per le due stampe
# ax è un array contenente due oggetti di tipo Axes (asse del grafico) 

fig, ax = plt.subplots(2, figsize = (8, 4) )
ax[0].plot(x, np.sin(x))    # Invochiamo plot() per stampare sul primo asse  
ax[1].plot(x, np.cos(x));   # Invochiamo plot() per stampare sul secondo asse
```

```python
plt.figure(figsize = (8, 3) )

plt.plot(x, np.sin(x), color='red')              # nome nel colore
plt.plot(x, np.sin(x + .5), color='b')           # codice corto del colore (rgbcmyk)
plt.plot(x, np.sin(x + 1), color='0.75')         # scala di grigio (da 0 a 1)
plt.plot(x, np.sin(x + 1.5), color='#FFDD44')    # Codice RGB esadecimale (da 0 a FF)
plt.plot(x, np.sin(x + 2), color=(1.0,0.2,0.3))  # Tupla RGB (da 0 a 1)
plt.plot(x, np.sin(x + 2.5), color='SteelBlue'); # Nome del colore (supportato un HTML)
```

```python
plt.figure(figsize = (8, 3) )

plt.plot(x, x , linestyle='solid')
plt.plot(x, x + 2, linestyle='dashed')
plt.plot(x, x + 4, linestyle='dashdot')
plt.plot(x, x + 6, linestyle='dotted');
```
